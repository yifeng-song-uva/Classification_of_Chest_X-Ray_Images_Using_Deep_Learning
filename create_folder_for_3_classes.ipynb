```python
import os
import shutil
import glob
import pandas as pd
from collections import Counter
```

```python
input_fpath =  r'C:\Users\songy\Downloads\UVA STAT 5630 Statistical Machine Learning\Project\data\chest_xray_resized_new'
output_fpath =  r'C:\Users\songy\Downloads\UVA STAT 5630 Statistical Machine Learning\Project\data\chest_xray_resized_new_3'
```

```python
try:
    os.mkdir(output_fpath)
except FileExistsError:
    pass
```

```python
file_paths_old = pd.read_csv(input_fpath + "/file_paths.csv")
```

```python
# create all directories and subdirectories
all_parts = Counter(file_paths_old.dataset)
all_classes = Counter(file_paths_old.sublabel)
for p in all_parts:
    try:
        os.mkdir(output_fpath + "/" + p)
    except FileExistsError:
        pass
    for c in all_classes:
        try:
            os.mkdir(output_fpath + "/" + p + "/" + c)
        except FileExistsError:
            pass
```

```python
def new_image_path(ip, sublabel):
    new_ip = ip.split("/")
    new_ip[0] += "_3"
    new_ip[2] = sublabel
    return "/".join(new_ip)

file_paths_old['image_path_new'] = file_paths_old.apply(lambda r: new_image_path(r.image_path, r.sublabel), axis=1)
```

```python
for i in range(file_paths_old.shape[0]):
    old_fp = file_paths_old.image_path.iloc[i].split("/")
    old_fp[0] = input_fpath
    old_fp = "/".join(old_fp)
    new_fp = file_paths_old.image_path_new.iloc[i].split("/")
    new_fp[0] = output_fpath
    new_fp = "/".join(new_fp)
    shutil.copy(old_fp, new_fp)
```

```python
file_paths_old.to_csv(output_fpath + "/file_paths.csv", index=False)
```
